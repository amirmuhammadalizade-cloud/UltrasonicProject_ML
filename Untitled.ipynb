```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

DATA_DIR = Path(r"C:\Users\asus\PycharmProjects\UltrasonicProject_ML")

data = np.load(
    DATA_DIR / "Pk050_3D_Dataset_Long_Rot00.npy",
    mmap_mode="r"
)

X = np.load(DATA_DIR / "X-values.npy")
Y = np.load(DATA_DIR / "Y-values.npy")
Z = np.load(DATA_DIR / "Z-values.npy")

# انتخاب یک نقطه
x_index = 100
y_index = 40

# استخراج A-scan
a_scan = data[x_index, y_index, :]

print("A-scan shape:", a_scan.shape)
print("X position:", X[x_index])
print("Y position:", Y[y_index])
print("Number of samples:", len(a_scan))

plt.figure(figsize=(12, 5))

plt.plot(Z, a_scan)

plt.xlabel("Z")
plt.ylabel("Amplitude")
plt.title(
    f"A-scan at X={X[x_index]}, Y={Y[y_index]}"
)

plt.grid()
plt.show()
```

```python
points = [
    (0, 0),
    (50, 20),
    (100, 40),
    (150, 60),
    (200, 80)
]

plt.figure(figsize=(12, 6))

for x_index, y_index in points:
    a_scan = data[x_index, y_index, :]

    plt.plot(
        Z,
        a_scan,
        label=f"X={X[x_index]}, Y={Y[y_index]}"
    )

plt.xlabel("Z")
plt.ylabel("Amplitude")
plt.title("A-scan comparison at different spatial positions")
plt.legend()
plt.grid()
plt.savefig("scan1.png")
plt.show()
```

```python
import numpy as np

# تمام A-scanها
# shape = (201, 81, 4000)

rms_map = np.sqrt(np.mean(data ** 2, axis=2))

print("RMS map shape:", rms_map.shape)
print("Minimum RMS:", rms_map.min())
print("Maximum RMS:", rms_map.max())
print("Mean RMS:", rms_map.mean())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.imshow(
    rms_map.T,
    origin="lower",
    aspect="auto",
    extent=[X.min(), X.max(), Y.min(), Y.max()]
)

plt.colorbar(label="RMS")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("RMS Map of A-scans")


plt.show()
```

```python
points = [
    (75, 2),
    (75, 4),
    (75, 5),
    (75, 6),
    (75, 7),
    (75, 8)
]

plt.figure(figsize=(12, 6))

alpha_index = 1
for x_index, y_index in points:
    a_scan = data[x_index, y_index, :]

    plt.plot(
        Z,
        a_scan,
        label=f"X={X[x_index]}, Y={Y[y_index]}",
        alpha= alpha_index

    )
    alpha_index -= 0.1

plt.xlabel("Z")
plt.ylabel("Amplitude")
plt.title("A-scan comparison at different spatial positions")
plt.legend()
plt.grid()

plt.savefig("scan3.png")
plt.show()
```
